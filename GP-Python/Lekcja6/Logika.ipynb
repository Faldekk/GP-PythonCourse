Napiszcie jakieś pętle z for i range

if sprawdza czy zdanie jest prawdziwe... 

Jeśli zdanie jest prawdziwe to wykonuje jakąś czynność, jeśli nie jest nie wykona tego albo wykona to co jest w else:...

```python
v = int(input("Podaj prędkość: "))
if v > 50:
    print("Jedziesz zbyt szybko! :( ")
else:
    print('Zwolnij')
```

Pewnie z matematyki znacie takie znaczki jak > czy < . Ale jeśli chcemy w pythonie sprawdzić czy coś jest sobie równe to używamy operatora = .

```python
haslo = 'masło123'
odp = input('Podaj hasło: ')

if odp == haslo:
    print('Poprawne hasło')
    print('Zalogowano')
    print('Tajne informacje')
else:
    print('Złe hasło')
```

Dobra teraz coś trudniejszego, czyli and. Wyrażenie A and B są prawdziwe tylko, gdy A i B jest prawdziwe

```python
login = 'maurycy12'
haslo = 'masło123'
user_login = input('Login: ')
user_pw = input('Hasło: ')

if user_login == login and user_pw == haslo:
    print('Zalogowano')
    print('Tajne informacje')
else:
    print('Niepoprawne dane')
```

To do kompletu dodajmy lub, czyli or. Wyrażenie A and B są prawdziwe, gdy A lub B jest prawdziwe

```python
import random

czas = random.randint(0, 10)
zycie = random.randint(-15, 15)

print(f'Życie: {zycie}')
print(f'Czas: {czas}')

if czas <= 0 or zycie <= 0:
    print('Game Over')
else:
    print('Gra się toczy')
```
