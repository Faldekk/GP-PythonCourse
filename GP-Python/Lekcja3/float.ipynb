dodawanie +
odejmowanie - 
dzielenie / 
mnozenie *
potęgowanie **

```python
liczba1 = 67
liczba2 = 12

print(f"{liczba1 / liczba2}") # dzielenie
print(f"{liczba1 // liczba2}") # dzielenie bez reszty
print(f"{liczba1 % liczba2}") # reszta z dzielenia
```

```python
x = 3

print(f"{(x+2)**(x+3)}")
print(pow(x+2,x+3))
```

Funckje wbudowane
round()

```python
print(f'{round(3.6)}')
print(f'{abs(-43)}')
print(f'{len("Hello World!")}') # length - długość
print(f'{max(-43,3,4,56,6,2,2,3)}')
print(f"{min(-43,3,4,56,6,2,2,-2)}")
```
