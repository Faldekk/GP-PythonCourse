```python
licz1 = 14
licz2 = 8
print(f"Dodawanie: {licz1 + licz2}") 
print(f"Odejmowanie: {licz1 - licz2}")
print(f"Mnożenie: {licz1 * licz2}")
print(f"Dzielenie: {licz1 / licz2}")
```

```python
# Try - spróbuj coś zrobić
# except - jeśli się nie uda
x = input("Podaj x: ")
try:
    x = int(x)
    print("Konwersja udana")
except:
    print("Nie udało się")
```

```python
# Kalkulator (+, -, /, *)
# bierzemy 2 zmienne
# przykładowy output: Dodawanie: 9 + 14 = 23
licz1 = 45
licz2 = 32
print(f"Dodawanie: {licz1} + {licz2} = {licz1 + licz2}") 
print(f"Odejmowanie: {licz1} - {licz2} = {licz1 - licz2}")
print(f"Mnożenie: {licz1} * {licz2} =  {licz1 * licz2}")
print(f"Dzielenie: {licz1} / {licz2} = {licz1 / licz2}")
```
