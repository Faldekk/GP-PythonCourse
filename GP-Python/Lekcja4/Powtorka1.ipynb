```python
# Pobierz odległość (kilometry) i czas (godzina) od użytkownika. Wylicz średnią prędkość
odleglosc = int(input("Podaj odległość: "))
czas = float(input("Podaj czas w godzinach: "))

print(f"Średnia prędkośc wynosi { odleglosc / czas }")
```

```python
# Pobierz przyprostokątne a i b. Oblicz przeciwprostokątną c. (Tw. Pitagorasa => a^2 + b^2 = c^2)
import math as m
a = int(input("Podaj a: "))
b = int(input("Podaj b: "))
print(f" {m.sqrt(a ** 2 + m.pow(b,2))}")
```

```python
# Parking ma 77 metrów długości. Jedno auto zajmuje 5 metrów, parkując równolegle. 
# Ile aut zmieści się na parkingu oraz ile miejsca pozostanie niewykorzystane?
parking = 77 
auto = 5
print(f"Ilość miejsc {parking // auto}")
print(f"Zostanie {parking % auto} niewykorzsytanego miejsca")
```

```python
# Pobierz od użytkownika liczbę - x. Wylosuj liczbę z przedziału (x, x+10) i wyświetl.
import random
x = int(input("Podaj x: "))
print(f"Wylosowana liczba {random.randint(x,x+10)}")
```

```python
# Wprowadzenie do list (Gry losowe i zapisywanie wyników)
import random

min_liczba = 1
max_liczba = 60

wylosowane_liczby = []

while len(wylosowane_liczby) < 6:
    liczba = random.randint(min_liczba, max_liczba)

    if liczba not in wylosowane_liczby:
        wylosowane_liczby.append(liczba)

wylosowane_liczby.sort() 

print("Wylosowane liczby:")
print(wylosowane_liczby)
```

Dobra to co to są te listy?
Są to po prostu koszyczki na zmienne. Jest to kolekcja wartości. 
Przykładowo: Mam listę kart, których wartość jest różna ale mam do nich przypisane ID - numery
tym samym mogę stworzyć listę z wartościami kart

```python
lista_zakupow = ["chleb", "ser", "mleko", "czekolada"]
```

Zamiana danych w liście po indeksie

```python
lista_zakupow[1] = "żółty ser" 
```

Dodawanie danych do listy (na końcu ląduje)

```python
lista_zakupow.append("rzodkiewki")
```

Usuwanie z listy po "nazwie" rzeczy

```python
lista_zakupow.remove("rzodkiewki")
```

Losowanie rzeczy z listy -> zawsze len() - 1 bo indeksy idą od 0

```python
import random as r

print(lista_zakupow[r.randint(0,len(lista_zakupow)-1)])
```

```python
print(len(lista_zakupow))
```
