Mamy jeszcze dodatkowe komendy do list 

max(lista) -> zwraca maksymalna wartość z listy

min(lista) -> zwraca minimalną wartość z listy

sum(lista) -> Sumuje wartośći w liście

```python
koszty = [159, 20.99, 13.21, 50]
print(f'Największy koszt: {max(koszty)} zł')
print(f'Najmniejszy koszt: {min(koszty)} zł')
suma = sum(koszty)
print(f'Suma kosztów: {suma} zł')
```

Będą momenty, gdzie jako programista nie będziemy mieli funkcji na wszystko.
W wielu językach jedną z podstaw w zrozumieniu języka są pętle. I tym się dzisiaj zajmiemy !!!

```python
kolory2 = ["Turkusowy", "Musztardowy", "Czarny", "Biały"]
for k in kolory2:
	print(f'Kolor: {k}')

koszt_farby = [31,43,56,70]

for kolor, koszt in zip(kolory, koszt_farby):  #zip(pozwala nam wypisać wiele wartości i przez nie naraz przechodzić)
    print(f"Cena farby koloru {kolor} wynosi {koszt} zł")
```
