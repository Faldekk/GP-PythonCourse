Mamy jeszcze dodatkowe komendy do list 

max(lista) -> zwraca maksymalna wartość z listy

min(lista) -> zwraca minimalną wartość z listy

sum(lista) -> Sumuje wartośći w liście

```python
koszty = [159, 20.99, 13.21, 50]
print(f'Największy koszt: {max(koszty)} zł')
print(f'Najmniejszy koszt: {min(koszty)} zł')
suma = sum(koszty)
print(f'Suma kosztów: {suma} zł')
```
